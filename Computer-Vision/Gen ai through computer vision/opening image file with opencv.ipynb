```python
import cv2

myimg = cv2.imread(r"C:\Users\LIKHITH\OneDrive\Pictures\child.jpg")


while True:
    cv2.imshow('nit', myimg )
    # if user waited for 1 ms and we have pressed esc key  
    if cv2.waitKey(1) & 0xFF == 27:
        break

cv2.destroyAllWindows()
```
