# array creation functions

```python
import numpy as np
```

```python
a= np.array([1,2,3])
print("array a:",a)
```

```python
b= np.arange(0,10,2)
print("array b:",b)
```

```python
d=np.zeros((2,3))
print(" array d:\n",d)
```

```python
d=np.zeros((2,3),dtype=int)
print(" array d:\n",d)
```

```python
f=np.eye(4)
print("Identity matrix f:\n",f)
```

# Array manipulation functions

```python
a1=np.array([1,2,3])
reshaped = np.reshape(a1,(1,3))
print("reshaped array:",reshaped)
```

```python
a1=np.array([1,2,3])
reshaped = np.reshape(a1,(3,1))
print("reshaped array:",reshaped)
```

```python
f1=np.array([[1,2],[3,4],[5,6],[7,8]])
flattend=np.ravel(f1)
print("flattend array:",flattend)
```

```python
e1=np.array([[2,4],[5,6]])
transposed=np.transpose(e1)
print("transposed array:\n",transposed)
```
