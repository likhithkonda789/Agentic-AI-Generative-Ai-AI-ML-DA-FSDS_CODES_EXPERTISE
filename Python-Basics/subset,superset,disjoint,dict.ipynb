# superset,subset,disjoint

```python
s4={1,2,3,4,5,6,7,8,9}
s5={3,4,5,6,7,8}
s6={10,20,30,40,5}
```

```python
s4.issuperset(s5)
```

```python
s4.issubset(s5)
```

```python
s5.issubset(s4)
```

```python
s5.isdisjoint(s6)
```

```python
s5.isdisjoint(s4)
```

```python
s5.issuperset(s6)
```

```python
s5.isdisjoint(s6)
```
